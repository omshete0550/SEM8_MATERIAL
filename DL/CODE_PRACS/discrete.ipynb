```python
import numpy as np
import itertools
```

```python
def unipolar(net):
    return 1 if (net >= 0) else 0
```

```python
def bipolar(net):
    return 1 if (net >= 0) else -1
```

```python
def predict(X, weights, type):
    net = np.dot(X, weights)
    if type == 'unipolar': 
        return unipolar(net)
    elif type == 'bipolar':
        return bipolar(net)
```
