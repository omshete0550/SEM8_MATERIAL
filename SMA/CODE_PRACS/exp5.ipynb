•	Positive Sentiment analysis


```python
import pandas as pd
from textblob import TextBlob
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("CSV/location.csv")
df.head()
```

```python
df['polarity'] = df['Text'].apply(lambda x : TextBlob(x).sentiment.polarity)
print(df.head())
```

```python
ans = []
for i in df["polarity"]:
  print(i)
  if i>0.5:
    ans.append("positive")
  else:
    ans.append("negative")

df["pred"] = ans
```

```python
print(df.head())
```

```python
pos_tweets = df[df["pred"] == "positive"]
neg_tweets = df[df["pred"] == "negative"]
```

```python
sns.kdeplot(pos_tweets['polarity'], shade=True, label='Positive')
sns.kdeplot(neg_tweets['polarity'], shade=True, label='Negative')
plt.title("Sentiment analysis of Tweets")
plt.xlabel("Polarity Score")
plt.title("Sentiment Analysis of Tweets")
plt.legend()
plt.show()
```

```python
ans = []
for i in df["polarity"]:
  if i>0.1:
    ans.append("positive")
  elif(i<-0.1):
    ans.append("negative")
  else:
    ans.append("neutral")

df["sentiment"] = ans
```

```python
sentiment_counts = df['sentiment'].value_counts()
print(sentiment_counts)
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values, palette='viridis')
plt.xlabel('Sentiment')
plt.ylabel('Number of Tweets')
plt.title('Sentiment Analysis')
plt.show()
```
