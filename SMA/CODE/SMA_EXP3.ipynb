```python
df_comments = pd.read_csv("/content/drive/MyDrive/SMA_DATASET/merged_comments.csv")
df_comments
```

```python
comments_list = df_comments["Comment"]
comments_list.head()
```

```python
lemmatizer = WordNetLemmatizer()
new_comment_list =[]
stopwords = ['the', 'of', 'and', 'is','to','in','a','from','by','that', 'with', 'this', 'as', 'an', 'are','its', 'at', 'for']
```

```python
for i in comments_list:
  # lower case
  str = i.lower()
  # remove symbols
  str = re.sub(r'[^\w]', ' ', str)
  # tokens conversion
  words_list = str.split(' ')
  # Lemmatization
  lemma_lst=[]
  for x in words_list:
    if len(x) >=2:
      lemma_lst.append(lemmatizer.lemmatize(x))
  # Removing stop words
  words_without_stopwords = [ word for word in lemma_lst if word not in stopwords ]
  new_comment_list.append(' '.join(words_without_stopwords))
  words_list = []
```

```python
for i in range(5):
  print(comments_list[i])
  print(new_comment_list[i],"\n\n")
```

```python
df_comments["Comment"] = new_comment_list
df_comments.head()
```
